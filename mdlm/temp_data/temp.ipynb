### 保存每个 strain 所需要的对应的 SELFIES 文件

```python
from rdkit import Chem
from transformers import AutoModel, AutoTokenizer
import selfies as sf
from tqdm import tqdm

# model_name = "ibm-research/materials.selfies-ted"
# tokenizer = AutoTokenizer.from_pretrained(model_name)
selfies_seqs = []
strains = ['19606', '11775', '13883', '47085', '#002', '12600', 'BAA-1556', '700802']
for peptide in tqdm(list(peptides), desc='Processing peptides'):
    mol = Chem.MolFromSequence(peptide.strip())
    try:
        smiles = Chem.MolToSmiles(mol)
    except:
        # print(peptide)
        continue
    # smiles = Chem.MolToSmiles(mol, canonical=True)
    selfies_seq = sf.encoder(smiles)
    selfies_seqs.append(selfies_seq)

for strain in strains:
    with open(f'strain_{strain}.txt', 'w') as f:
        f.write('\n'.join(selfies_seqs))
```
