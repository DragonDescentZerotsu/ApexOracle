#### 去重处理 uniprot 和 uniref 中的 peptide

```python
import sys
from Bio import SeqIO  # Biopython's module for sequence I/O
from tqdm import tqdm  # Progress bar library

# --- 配置 ---
# 将下面替换成你的实际文件名
fasta_files = ["/data1/tianang/SMILES_data/uniprotkb_length_1_TO_50_2025_04_09.fasta", "/data1/tianang/SMILES_data/uniref_length_1_TO_50_2025_04_03.fasta"]
output_file = "./Data/unique_peptide_sequences.txt"

# --- 主逻辑 ---
print("开始处理 FASTA 文件...")

unique_sequences = set() # 使用集合来自动处理重复项

# 遍历输入文件列表
for filename in fasta_files:
    print(f"\n正在处理文件: {filename}")
    try:
        # 使用 SeqIO.parse 读取 FASTA 文件，它返回一个 SeqRecord 迭代器
        # 使用 tqdm 包装迭代器以显示进度条
        # 注意：tqdm 可能无法预先知道 FASTA 文件中的确切记录数，
        # 但它仍然会显示处理的记录数和速率。
        # 使用 unit='records' 使进度条显示单位更清晰
        # record_iterator = SeqIO.parse(filename, "fasta")

        # 为了让tqdm知道总数（可选，但能让进度条更好看），可以先计数
        # 但这会读取文件两次，对于大文件可能效率不高。
        # 更常见的方法是直接迭代，tqdm会动态显示进度。
        # 我们这里直接迭代：

        # 使用 with 来确保文件句柄被正确关闭（虽然SeqIO通常会处理好）
        with open(filename, 'r') as handle:
             # 使用 tqdm 包装 SeqIO.parse
             # desc 参数为进度条添加描述性文字
            for record in tqdm(SeqIO.parse(handle, "fasta"), desc=f"读取 {filename}", unit=" 条记录"):
                # record.seq 是一个 Seq 对象，需要转换为字符串添加到集合中
                sequence_str = str(record.seq)
                unique_sequences.add(sequence_str)

    except FileNotFoundError:
        print(f"\n错误：输入文件 '{filename}' 未找到。", file=sys.stderr)
        print("请检查文件名和路径是否正确。")
        sys.exit(1) # 退出程序
    except Exception as e:
        print(f"\n处理文件 '{filename}' 时发生错误: {e}", file=sys.stderr)
        sys.exit(1) # 退出程序

total_unique_count = len(unique_sequences)
print(f"\n在所有文件中总共找到 {total_unique_count} 条独特的蛋白质序列。")

# 将独特的序列写入输出文件，并为写入过程添加进度条
try:
    print(f"\n正在将结果写入文件: {output_file}")
    with open(output_file, 'w') as outfile:
        # 包装 unique_sequences 进行迭代，以显示写入进度
        for seq in tqdm(unique_sequences, desc="写入文件", unit=" 条序列", total=total_unique_count):
            outfile.write(seq + '\n') # 每条序列写一行
    print(f"\n成功将 {total_unique_count} 条独特序列写入 {output_file}。")
    print("处理完成。")
except Exception as e:
    print(f"\n写入输出文件 '{output_file}' 时发生错误: {e}", file=sys.stderr)
    sys.exit(1)
```

#### 把这些 peptide sequence 全部转换成 SMILES string
*注：实际上由 MDLM / uniprot_uniref_to_smiles.py 的并行版本完成

```python
from tqdm import tqdm
from aa_seq_to_smiles import *
from rdkit import Chem
import json

# --- 文件配置 ---
input_peptide_file = "/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/unique_peptide_sequences.txt"  # 包含肽序列的输入文件
output_smiles_file = "/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/unique_peptides_as_smiles.txt" # 输出SMILES的文件
aa_smiles_dict = get_aa_smiles_dict('/data1/tianang/Projects/Synergy/DataPrepare/Data/all_aa_smiles_new_handcrafted.csv')

# 标准氨基酸集合（兼容大小写检测）
canonical_aas = {'A', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'K', 'L',
                'M', 'N', 'P', 'Q', 'R', 'S', 'T', 'V', 'W', 'Y',
                'a', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'k', 'l',
                'm', 'n', 'p', 'q', 'r', 's', 't', 'v', 'w', 'y'}

def contains_non_canonical(peptide):
    """检测序列中是否包含非标准氨基酸（兼容大小写）"""
    return any(aa not in canonical_aas for aa in peptide)

print(f"正在从 '{input_peptide_file}' 读取肽序列...")

# 计算总行数
with open(input_peptide_file, 'r') as f:
    total_lines = sum(1 for _ in f)

print(f"总共有 {total_lines} 条肽序列需要处理。")
print(f"正在将SMILES写入 '{output_smiles_file}'...")

invalid_count = 0

with open(input_peptide_file, 'r') as infile, open(output_smiles_file, 'w') as outfile:
    line_num = 0

    for line in tqdm(infile, total=total_lines, desc="处理进度"):
        line_num += 1
        raw_peptide = line.strip()  # 保留原始大小写

        # 跳过空行
        if not raw_peptide:
            continue

        # 检测非标准氨基酸（直接使用原始大小写检测）
        if contains_non_canonical(raw_peptide):
            print(f"行号 {line_num}: 发现非标准氨基酸 - {raw_peptide}")
            invalid_count += 1
            continue

        # 转换并写入（使用原始序列）
        pep_obj = Peptide(raw_peptide, aa_smiles_dict=aa_smiles_dict)
        smiles_result = Chem.MolToSmiles(pep_obj.ncTerminus_modified_mols[0])
        if smiles_result:
            outfile.write(smiles_result + '\n')

# 最终统计报告
print(f"\n处理完成。共过滤 {invalid_count} 条含非标准氨基酸的序列。")
print(f"有效肽序列数量：{total_lines - invalid_count}")
print(f"无效肽序列数量：{invalid_count}")
```

#### 把 all_smiles.csv 中不同来源的 smiles 单独存一下

```python
import pandas as pd
import numpy as np
from tqdm import tqdm


def split_smiles_by_source(input_file):
    # 读取数据为NumPy数组
    df = pd.read_csv(input_file)
    data = df.values  # 获取二维数组 [N rows x 2 cols]

    # 创建字典存储不同来源的数据
    source_dict = {}

    # 逐行处理数据
    for row in tqdm(data, desc="Splitting SMILES", unit=" lines"):
        original_id, smiles = row
        source_name = original_id.rsplit('_', 1)[0]
        # try:
            # 拆分ID获取来源名称
            # source_name = original_id.rsplit('_', 1)[0]
            # if len(parts) != 2:  # 跳过无效ID
            #     continue
            # source_name, _ = parts
        # except:
        #     continue

        # 存储数据（保持原始ID）
        if source_name not in source_dict:
            source_dict[source_name] = []
        source_dict[source_name].append([original_id, smiles])

    # 保存每个来源的数据
    for source_name, rows in source_dict.items():
        # 转换为DataFrame并保存
        pd.DataFrame(
            rows,
            columns=['ID', 'SMILES']
        ).to_csv(f"/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/{source_name}_smiles.csv", index=False)


if __name__ == "__main__":
    split_smiles_by_source("/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/all_smiles.csv")
```

#### 筛掉 pubchem 里面所有的有肽键的分子

```python
import pandas as pd

# 1. 读入 CSV
df = pd.read_csv('/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/pubchem_smiles.csv')  # 把文件名替换成你的

# 2. 构造布尔掩码：SMILES 中既不含 "NC(=O)" 也不含 "C(=O)N"
mask = ~(
    df['SMILES'].str.contains('NC(=O)', regex=False) |
    df['SMILES'].str.contains('C(=O)N', regex=False)
)

# 3. 只保留 mask 为 True 的行
df_filtered = df[mask]

print(f' original length: {len(df)}')
print(f' filtered length: {len(df_filtered)}')

# 4. 写出到新文件
df_filtered.to_csv('/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/pubchem_no_peptide.csv', index=False)
```

#### 造一个新的从 pubchem 中过滤掉多肽的 数据集用于训练分类器

```python
import pandas as pd
from tqdm import tqdm

# 1. 准备文件列表和对应的 label
#    这里把 pos.csv 标记为 1，其他文件标记为 0
files = {
    '/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/SmProt2_smiles.csv': 1,
    '/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/pubchem_no_peptide.csv': 0,
    '/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/Generated_pep_CLM_smiles.csv': 1,
    '/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/uni_smiles.csv': 1
    # … 如果还有更多文件，继续加在这里
}

# 2. 依次读取、打 label、收集到列表
dfs = []
for fname, lbl in tqdm(files.items()):
    df = pd.read_csv(fname)     # 读取 ID, SMILES 两列
    df['label'] = lbl           # 添加新列
    dfs.append(df)

# 3. 拼接所有 DataFrame，并重置索引
df_all = pd.concat(dfs, ignore_index=True)

# 4. 保存到新的 CSV
print(f'saving')
df_all.to_csv('/data1/tianang/Projects/Synergy/DataPrepare/MDLM/Data/all_smiles_pep_SM_cls.csv', index=False)
```
